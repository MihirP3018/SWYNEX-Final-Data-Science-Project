# ***Task 1:- Data Preparation***

#### ***Import Libraries***

```python
import numpy as np
import pandas as pd
```

#### ***Reading The Dataset***

```python
df = pd.read_csv('smartphones_uncleaned.csv')
```

#### ***Check The Shape (Rows And Columns) Of The Dataset***

```python
df.shape
```

#### ***Check The First 5 information Of The Dataset***

```python
df.head()
```

#### ***Check The Information Of Dataset***

```python
df.info()
```

#### ***Check missing values***

```python
df.isnull().sum()
```

#### ***Check duplicate rows***

```python
df.duplicated().sum()
```

#### ***Remove duplicate rows***

```python
df = df.drop_duplicates()
```

```python
print("Shape after removing duplicates:", df.shape)
```

#### ***Handle missing values***

```python
numric_cols = df.select_dtypes(include = np.number).columns
categorical_cols = df.select_dtypes(include = 'object').columns
```

#### ***Fill numerical missing values with median***

```python
for column in numric_cols:
    df[column] = df[column].fillna(df[column].median())
```

#### ***Fill categorical missing values with mode***

```python
for column in categorical_cols:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].mode()[0])
```

#### ***Check missing values again***

```python
print("Missing values after handling:")
print(df.isnull().sum().sum())
```

#### ***Check unique values in categorical columns***

```python
for column in categorical_cols:
    print(f'\n{column}:')
    print(df[column].value_counts().head(10))
```

#### ***Check numerical summary***

```python
df.describe()
```

#### ***Check final duplicates***

```python
print("\nDuplicates after cleaning:", df.duplicated().sum())
```

#### ***Final dataset information***

```python
print("Final Dataset Shape:", df.shape)
print("Final Missing Values:", df.isnull().sum().sum())
```

#### ***Save prepared dataset***

```python
df.to_csv("smartphones_prepared.csv", index=False)
```
